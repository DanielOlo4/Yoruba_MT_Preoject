```python
"""English -> Yorùbá tagged lexicon for the rule-based MT project.

Each English word maps to a POS tag and its Yorùbá equivalent.
Lines marked REVIEW need a check from a Yorùbá speaker.
Lines marked ADDED were not in the original 100-word list.
"""
import unicodedata


def nfc(text):
    """Store every Yorùbá string in one Unicode form so tone marks always match."""
    return unicodedata.normalize("NFC", text)



ENTRIES = [
    # People
    ("person", "N", "ẹni"),
    ("child", "N", "ọmọ"),
    ("king", "N", "ọba"),
    ("father", "N", "bàbá"),
    ("mother", "N", "ìyá"),
    ("brother", "N", "arákùnrin"),
    ("sister", "N", "arábìnrin"),
    ("friend", "N", "ọ̀rẹ́"),
    ("teacher", "N", "olùkọ́"),
    ("student", "N", "akẹ́kọ̀ọ́"),

    # Places and transport
    ("house", "N", "ilé"),
    ("school", "N", "ilé-ẹ̀kọ́"),
    ("market", "N", "ọjà"),
    ("vehicle", "N", "ọkọ̀"),
    ("car", "N", "ọkọ̀ ayọ́kẹ́lẹ́"),
    ("train", "N", "ọkọ̀ ojú irin"),

    # Food and things
    ("water", "N", "omi"),
    ("food", "N", "oúnjẹ"),
    ("bread", "N", "bùrẹ́dì"),
    ("pounded yam", "N", "iyán"),  
    ("meat", "N", "ẹran"),
    ("fish", "N", "ẹja"),
    ("salt", "N", "iyọ̀"),
    ("money", "N", "owó"),
    ("cloth", "N", "aṣọ"),
    ("shoe", "N", "bàtà"),
    ("book", "N", "ìwé"),
    ("pencil", "N", "pẹ́ǹsù"),
    ("table", "N", "tábìlì"),
    ("chair", "N", "àga"),
    ("door", "N", "ilẹ̀kùn"),
    ("window", "N", "fèrèsé"),
    ("rice", "N", "ìrẹsì"),  

    # Body
    ("eye", "N", "ojú"),
    ("hand", "N", "ọwọ́"),
    ("head", "N", "orí"),

    # Time
    ("day", "N", "ọjọ́"),
    ("night", "N", "alẹ́"),
    ("morning", "N", "òwúrọ̀"),
    ("evening", "N", "ìrọ̀lẹ́"),
    ("today", "ADV", "òní"),
    ("tomorrow", "ADV", "ọ̀la"),
    ("yesterday", "ADV", "àná"),
    ("there", "ADV", "nibẹ̀"),

    # Adjectives
    ("big", "ADJ", "ńlá"),
    ("small", "ADJ", "kéré"),
    ("good", "ADJ", "dára"),
    ("bad", "ADJ", "búburú"),
    ("new", "ADJ", "tuntun"),
    ("old", "ADJ", "àtijọ́"),
    ("red", "ADJ", "pupa"),
    ("white", "ADJ", "funfun"),
    ("black", "ADJ", "dúdú"),
    ("green", "ADJ", "aláwọ̀ ewé"),
    ("blue", "ADJ", "búlúù"),
    ("yellow", "ADJ", "ofeefee"),

    # Verbs (base forms; the analysis step reduces eating/ate/eats to eat)
    ("want", "V", "fẹ́"),
    ("eat", "V", "jẹ"),
    ("drink", "V", "mu"),
    ("go", "V", "lọ"),
    ("come", "V", "wá"),
    ("see", "V", "rí"),
    ("hear", "V", "gbọ́"),
    ("say", "V", "sọ"),
    ("read", "V", "ka"),
    ("write", "V", "kọ"),
    ("sleep", "V", "sùn"),
    ("stand", "V", "dúró"),
    ("sit", "V", "jókòó"),
    ("buy", "V", "ra"),
    ("sell", "V", "tà"),
    ("receive", "V", "gbà"),
    ("do", "V", "ṣe"),
    ("know", "V", "mọ̀"),  
    ("learn", "V", "kọ́"),
    ("teach", "V", "kọ́ni"),
    ("help", "V", "rànlọ́wọ́"), 
    ("play", "V", "ṣeré"),

    # Negation
    ("not", "NEG", "kò"),

    # Prepositions
    ("to", "PREP", "sí"),
    ("in", "PREP", "ní"),
    ("at", "PREP", "ní"),
    ("with", "PREP", "pẹ̀lú"),
    ("from", "PREP", "láti"),
    

    # Conjunctions
    ("and", "CONJ", "àti"),
    ("but", "CONJ", "ṣùgbọ́n"),
    ("because", "CONJ", "nítorí"),

    # Determiners and possessives (both go AFTER the noun in Yorùbá)
    ("the", "DET", "náà"),   #náà only for known referents
    ("a", "DET", "kan"),  
    ("my", "POSS", "mi"),  
]

# Pronouns need separate subject and object forms.
# None = no fixed form yet (3sg object copies the verb's final vowel).
PRONOUNS = {
    "i":       {"pos": "PRON", "subj": "mo",  "obj": "mi"},
    "you":     {"pos": "PRON", "subj": "o",   "obj": "ẹ"},   # REVIEW: ó is the emphatic form
    "he":      {"pos": "PRON", "subj": "ó",   "obj": None},
    "she":     {"pos": "PRON", "subj": "ó",   "obj": None},
    "it":      {"pos": "PRON", "subj": "ó",   "obj": None},
    "we":      {"pos": "PRON", "subj": "a",   "obj": "wa"},  # REVIEW: àwa is the emphatic form
    "you all": {"pos": "PRON", "subj": "ẹ",   "obj": "yín"}, # REVIEW: ẹ̀yin is the emphatic form
    "they":    {"pos": "PRON", "subj": "wọ́n", "obj": "wọn"},
}

# Markers inserted by transfer rules. English has no word for these,
# so they are keyed by the tag the analysis step finds.
MARKERS = {
    "PROG": "ń",
    "PERF": "ti",
    "FUT": "máa",
    "FUT_2": "yóò",
    "FUT_3": "á",
    "QM": "ṣé",
    "QM_2": "ǹjẹ́",
}


def build_lexicon():
    lexicon = {}
    for english, pos, yoruba in ENTRIES:
        if english in lexicon:
            raise ValueError(f"Duplicate English entry: {english}")
        lexicon[english] = {"pos": pos, "yo": nfc(yoruba)}
    for english, entry in PRONOUNS.items():
        if english in lexicon:
            raise ValueError(f"Duplicate English entry: {english}")
        lexicon[english] = {
            "pos": entry["pos"],
            "subj": nfc(entry["subj"]),
            "obj": nfc(entry["obj"]) if entry["obj"] else None,
        }
    return lexicon


LEXICON = build_lexicon()
MARKERS = {tag: nfc(form) for tag, form in MARKERS.items()}

# Longest English entries first, so "pounded yam" is matched before "pounded" or "yam".
MULTIWORD_KEYS = sorted((k for k in LEXICON if " " in k), key=len, reverse=True)


if __name__ == "__main__":
    print(f"{len(LEXICON)} entries loaded, no duplicates.")
    print("Multiword English entries:", MULTIWORD_KEYS)
    
```
